# Logic Gates Implementation in Python

This notebook contains implementation of:
- AND Gate
- OR Gate
- XOR Gate

using Python functions and truth tables.


```python
# AND Gate Implementation

def AND_gate(a, b):
    return a and b

print("AND Gate Truth Table")
print("A B Output")

for a in [0, 1]:
    for b in [0, 1]:
        print(a, b, AND_gate(a, b))
```

```python
# OR Gate Implementation

def OR_gate(a, b):
    return a or b

print("OR Gate Truth Table")
print("A B Output")

for a in [0, 1]:
    for b in [0, 1]:
        print(a, b, OR_gate(a, b))
```

```python
# XOR Gate Implementation

def XOR_gate(a, b):
    return a ^ b

print("XOR Gate Truth Table")
print("A B Output")

for a in [0, 1]:
    for b in [0, 1]:
        print(a, b, XOR_gate(a, b))
```
